```python
import sympy
from sympy.physics.quantum.fermion import FermionOp
from sympy.physics.quantum import Dagger
from sympy.physics.quantum.operatorordering import normal_ordered_form

from functools import reduce
import numpy as np

from codes.second_approach import many_body_matrix
```

```python
# define symbols
symbols = sympy.symbols(
    r"A, U, \mu_L, \mu_R, t",
    real=True,
    commutative=True,
    positive=True
)
A, U, mu_L, mu_R, t = symbols

# define fermionic operators
c_L_u, c_L_d = FermionOp(r'c_{L, \uparrow}'), FermionOp(r'c_{L, \downarrow}')
c_R_u, c_R_d = FermionOp(r'c_{R, \uparrow}'), FermionOp(r'c_{R, \downarrow}')
```

```python
H = A #constant term
H += mu_L * Dagger(c_L_d) * c_L_d + mu_R * Dagger(c_R_d) * c_R_d + t * Dagger(c_L_d) * c_R_d + t * Dagger(c_R_d) * c_L_d # normal ham
H += U * Dagger(c_L_d) * c_L_d * Dagger(c_R_d) * c_R_d 
```

```python
%%time
ham = many_body_matrix(symbolic_ham=H, basis=[c_L_d, c_R_d])
```
